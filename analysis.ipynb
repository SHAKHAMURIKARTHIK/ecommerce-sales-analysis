# E-commerce Sales Analysis

Beginner walkthrough: inspect, clean, summarize, and visualize the sales dataset.

## 1. Import Pandas and load the data
`head()` displays the first five records.

```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_FOLDER = Path.cwd()
DATA_FILE = PROJECT_FOLDER / "data" / "sales_data.csv"
df = pd.read_csv(DATA_FILE)
df.head()
```

## 2. Explore the dataset
`shape` gives rows and columns; `info` shows data types; `isnull().sum()` counts missing values.

```
print("Rows and columns:", df.shape)
print("Column names:", df.columns.tolist())
df.info()
print("Missing values per column:\n", df.isnull().sum())
```

## 3. Clean the data

```
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
for column in ["Quantity", "Price", "Total_Sales"]:
    df[column] = pd.to_numeric(df[column], errors="coerce")
df = df.dropna().drop_duplicates()
df = df[(df["Quantity"] > 0) & (df["Price"] >= 0) & (df["Total_Sales"] >= 0)]
df["Month"] = df["Date"].dt.to_period("M").astype(str)
print("Cleaned shape:", df.shape)
df.head()
```

## 4. Analyze sales using groupby

```
product_sales = df.groupby("Product", as_index=False)["Total_Sales"].sum().sort_values("Total_Sales", ascending=False)
monthly_sales = df.groupby("Month", as_index=False)["Total_Sales"].sum().sort_values("Month")
region_sales = df.groupby("Region", as_index=False)["Total_Sales"].sum().sort_values("Total_Sales", ascending=False)
product_sales, monthly_sales, region_sales
```

## 5. Visualize results

```
plt.figure(figsize=(9, 5))
plt.bar(product_sales["Product"], product_sales["Total_Sales"])
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```
plt.figure(figsize=(9, 5))
plt.plot(monthly_sales["Month"], monthly_sales["Total_Sales"], marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=25)
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```
plt.figure(figsize=(8, 5))
plt.pie(region_sales["Total_Sales"], labels=region_sales["Region"], autopct="%1.1f%%", startangle=90)
plt.title("Sales Distribution by Region")
plt.tight_layout()
plt.show()
```
